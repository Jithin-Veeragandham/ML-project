```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import pandas as pd
from tqdm import tqdm
from torch.utils.data import DataLoader, TensorDataset


X_train_tensor = torch.tensor(X_train.values, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train.values, dtype=torch.float32).view(-1, 1)
X_test_tensor = torch.tensor(X_valid.values, dtype=torch.float32)
y_test_tensor = torch.tensor(y_valid.values, dtype=torch.float32).view(-1, 1)

class MLP(nn.Module):
    def __init__(self):
        super(MLP, self).__init__()
        self.fc1 = nn.Linear(X_train.shape[1], 64)  
        self.fc2 = nn.Linear(64, 1) 
        self.relu = nn.ReLU()
        self.sigmoid = nn.Sigmoid()
 
    def forward(self, x):
        x = self.relu(self.fc1(x))
        x = self.sigmoid(self.fc2(x))
        return x
 
model = MLP()
if torch.cuda.is_available():
    device = torch.device("cuda")
    model.to(device)
    X_train_tensor = X_train_tensor.to(device)
    y_train_tensor = y_train_tensor.to(device)
    X_test_tensor = X_test_tensor.to(device)
    y_test_tensor = y_test_tensor.to(device)
else:
    device = torch.device("cpu")
    print("CUDA is not available. Using CPU.")

# Loss and optimizer
criterion = nn.BCELoss()  # Binary Cross-Entropy loss
optimizer = optim.Adam(model.parameters(),lr=1e-5, weight_decay=1e-4)  # Try values like 1e-4, 1e-5

train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
test_dataset = TensorDataset(X_test_tensor, y_test_tensor)
batch_size = 2048
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=2)  # Reduce patience

best_test_loss = float('inf')
best_model_weights = None

epochs = 50
for epoch in range(epochs):
    current_lr = optimizer.param_groups[0]['lr']
    model.train()
    total_loss = 0
    train_correct = 0
    train_total = 0
    
    with tqdm(train_loader, desc=f'Epoch {epoch+1}/{epochs} - LR: {current_lr:.6f}') as pbar:
        for X_batch, y_batch in pbar:
            optimizer.zero_grad()
            outputs = model(X_batch)
            loss = criterion(outputs, y_batch)
            
            predictions = (outputs > 0.5).float()
            batch_correct = (predictions == y_batch).sum().item()
            batch_total = y_batch.numel()
            batch_acc = (batch_correct / batch_total) * 100
            
            train_correct += batch_correct
            train_total += batch_total
            total_loss += loss.item()
            
            loss.backward()
            optimizer.step()
            
            pbar.set_postfix(loss=f"{loss.item():.4f}", acc=f"{batch_acc:.2f}%")
    
    avg_loss = total_loss / len(train_loader)
    train_accuracy = train_correct / train_total
    print(f"Epoch {epoch+1} Loss: {avg_loss:.4f}, Train Accuracy: {train_accuracy*100:.2f}%")

    # Evaluate on test data
    model.eval()
    test_correct = 0
    test_total = 0
    test_loss = 0

    with torch.no_grad():
        for X_batch, y_batch in test_loader:  
            outputs = model(X_batch)
            loss = criterion(outputs, y_batch)
            test_loss += loss.item()
            
            predictions = (outputs > 0.5).float()
            batch_correct = (predictions == y_batch).sum().item()
            batch_total = y_batch.numel()
            test_correct += batch_correct
            test_total += batch_total
    
    test_loss /= len(test_loader)
    test_accuracy = test_correct / test_total
    print(f"Test Loss: {test_loss:.4f}, Test Accuracy: {test_accuracy*100:.2f}%")

    # Check if this is the best model so far
    if test_loss < best_test_loss:
        best_test_loss = test_loss
        best_model_weights = model.state_dict().copy()  # Save a copy of the current weights
    
    # Reduce LR if validation loss doesn't improve
    scheduler.step(test_loss)
    
    # Print updated learning rate
    new_lr = optimizer.param_groups[0]['lr']
    print(f"Learning Rate: {new_lr:.6f}\n")

```

```python
epochs = 50
for epoch in range(epochs):
    current_lr = optimizer.param_groups[0]['lr']
    model.train()
    total_loss = 0
    train_correct = 0
    train_total = 0
    
    with tqdm(train_loader, desc=f'Epoch {epoch+1}/{epochs} - LR: {current_lr:.6f}') as pbar:
        for X_batch, y_batch in pbar:
            optimizer.zero_grad()
            outputs = model(X_batch)
            loss = criterion(outputs, y_batch)
            
            predictions = (outputs > 0.5).float()
            batch_correct = (predictions == y_batch).sum().item()
            batch_total = y_batch.numel()
            batch_acc = (batch_correct / batch_total) * 100
            
            train_correct += batch_correct
            train_total += batch_total
            total_loss += loss.item()
            
            loss.backward()
            optimizer.step()
            
            pbar.set_postfix(loss=f"{loss.item():.4f}", acc=f"{batch_acc:.2f}%")
    
    avg_loss = total_loss / len(train_loader)
    train_accuracy = train_correct / train_total
    print(f"Epoch {epoch+1} Loss: {avg_loss:.4f}, Train Accuracy: {train_accuracy*100:.2f}%")

    # Evaluate on test data
    model.eval()
    test_correct = 0
    test_total = 0
    test_loss = 0

    with torch.no_grad():
        for X_batch, y_batch in test_loader:  
            outputs = model(X_batch)
            loss = criterion(outputs, y_batch)
            test_loss += loss.item()
            
            predictions = (outputs > 0.5).float()
            batch_correct = (predictions == y_batch).sum().item()
            batch_total = y_batch.numel()
            test_correct += batch_correct
            test_total += batch_total
    
    test_loss /= len(test_loader)
    test_accuracy = test_correct / test_total
    print(f"Test Loss: {test_loss:.4f}, Test Accuracy: {test_accuracy*100:.2f}%")

    # Check if this is the best model so far
    if test_loss < best_test_loss:
        best_test_loss = test_loss
        best_model_weights = model.state_dict().copy()  # Save a copy of the current weights
    
    # Reduce LR if validation loss doesn't improve
    scheduler.step(test_loss)
    
    # Print updated learning rate
    new_lr = optimizer.param_groups[0]['lr']
    print(f"Learning Rate: {new_lr:.6f}\n")

```
