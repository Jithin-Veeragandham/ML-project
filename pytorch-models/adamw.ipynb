```python
# Import statements

import time
import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd
from tqdm import tqdm
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
# Define location of training and testing data

train_dir='/Users/dragonfruit/cs6140-ml-workspace/ML-project/data/train_data_scaled_encoded_shrunk.csv'
test_dir='/Users/dragonfruit/cs6140-ml-workspace/ML-project/data/validation_data_scaled_encoded_shrunk.csv'
```

```python
# Load training and testing data

train=pd.read_csv(train_dir, dtype=train_dtypes)
X_train_scaled = train.drop(columns=['HasDetections'])
y_train = train['HasDetections']

del train

test=pd.read_csv(test_dir, dtype=train_dtypes)
X_test_scaled = test.drop(columns=['HasDetections'])
y_test = test['HasDetections']

del test
```

```python
# Convert to PyTorch tensors
X_train_tensor = torch.tensor(X_train_scaled.values, dtype=torch.float32)
X_test_tensor = torch.tensor(X_test_scaled.values, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train.values, dtype=torch.float32).view(-1, 1)
y_test_tensor = torch.tensor(y_test.values, dtype=torch.float32).view(-1, 1)

# Create datasets and data loaders
train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
test_dataset = TensorDataset(X_test_tensor, y_test_tensor)

train_loader = DataLoader(train_dataset, batch_size=1024, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=1024, shuffle=False)

# Define model
class BinaryClassifier(nn.Module):
    def __init__(self, input_dim):
        super(BinaryClassifier, self).__init__()
        self.layers = nn.Sequential(
            nn.Linear(input_dim, 512),
            nn.ReLU(),
            nn.BatchNorm1d(512),
            nn.Dropout(0.3),

            nn.Linear(512, 256),
            nn.ReLU(),
            nn.BatchNorm1d(256),
            nn.Dropout(0.3),

            nn.Linear(256, 128),
            nn.ReLU(),
            nn.BatchNorm1d(128),
            nn.Dropout(0.4),

            nn.Linear(128, 64),
            nn.ReLU(),
            nn.BatchNorm1d(64),
            nn.Dropout(0.4),

            nn.Linear(64, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.layers(x)

input_dim = X_train_tensor.shape[1]
model = BinaryClassifier(input_dim)

# Use AdamW optimizer
optimizer = optim.AdamW(model.parameters(), lr=0.001, weight_decay=1e-5)
criterion = nn.BCELoss()

# Learning rate scheduler
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5, min_lr=1e-6)

# Epochs
epochs = 25

# MPS Usage
if torch.backends.mps.is_available():
    mps_device = torch.device("mps")
    X_train_tensor = X_train_tensor.to(mps_device)
    X_test_tensor = X_test_tensor.to(mps_device)
    y_train_tensor = y_train_tensor.to(mps_device)
    y_test_tensor = y_test_tensor.to(mps_device)
    model.to(mps_device)

    train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True, pin_memory=True)
    test_loader = DataLoader(test_dataset, batch_size=512, shuffle=False, pin_memory=True)

    epochs = 30
    for epoch in tqdm(range(epochs), desc="Epochs (MPS)"):
        model.train()
        # Initialize training loss
        total_train_loss = 0.0
        for inputs, targets in train_loader:
            inputs, targets = inputs.to(mps_device), targets.to(mps_device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            total_train_loss += loss.item()
            loss.backward()
            optimizer.step()
            # Delete after use
            del inputs, targets, loss
        # Calculate the average training loss
        average_train_loss = total_train_loss / len(train_loader)

        model.eval()
        val_loss = 0.0
        with torch.no_grad():
            for inputs, targets in test_loader:
                inputs, targets = inputs.to(mps_device), targets.to(mps_device)
                outputs = model(inputs)
                val_loss += criterion(outputs, targets).item()
                # Delete after use
                del inputs, targets
        val_loss /= len(test_loader)
        scheduler.step(val_loss)

        # Print learning rate
        current_lr = optimizer.param_groups[0]['lr']
        print(f'Epoch {epoch+1}/{epochs}, Training Loss: {average_train_loss:.4f}, Validation Loss: {val_loss:.4f}, Learning Rate: {current_lr:.6f}')

        # Print accuracies every other epoch
        if (epoch + 1) % 2 == 0:
            train_accuracies = []
            test_accuracies = []
    
            model.eval()
            with torch.no_grad():
                # Process train data in batches
                for inputs, targets in DataLoader(train_dataset, batch_size=1024): #adjust batch size as needed
                    inputs = inputs.to(mps_device)
                    outputs = model(inputs)
                    preds = (outputs > 0.5).int().cpu().numpy()
                    targets = targets.cpu().numpy()
                    train_accuracies.append(accuracy_score(targets, preds))
                    del inputs, outputs, preds, targets
    
                # Process test data in batches
                for inputs, targets in DataLoader(test_dataset, batch_size=1024): #adjust batch size as needed
                    inputs = inputs.to(mps_device)
                    outputs = model(inputs)
                    preds = (outputs > 0.5).int().cpu().numpy()
                    targets = targets.cpu().numpy()
                    test_accuracies.append(accuracy_score(targets, preds))
                    del inputs, outputs, preds, targets
    
            train_accuracy = sum(train_accuracies) / len(train_accuracies)
            test_accuracy = sum(test_accuracies) / len(test_accuracies)
    
            print(f"Train Accuracy in epoch {epoch+1}: {train_accuracy:.4f}, Test Accuracy in epoch {epoch+1}: {test_accuracy:.4f}")
        
        if (epoch + 1) % 10 == 0 or (epoch + 1) % epochs == 0:
            # Every 10 epochs or on the final epoch, save the trained model weights after training 
            torch.save(model.state_dict(), f'trained_baseline_model_mps_weights_epoch_{epoch+1}.pth')
            print(f"Trained model weights saved to trained_baseline_model_mps_weights_epoch_{epoch+1}.pth")

else:
    print("MPS is not available.")
```
